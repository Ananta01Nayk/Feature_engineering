# COlumn Transform

```python
#import require library
import pandas as pd
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.preprocessing import OrdinalEncoder
from sklearn.impute import SimpleImputer
from sklearn.model_selection import train_test_split
```

```python
# load the data set
data = pd.read_csv('covid_toy.csv')
```

```python
data.sample(5)
```

```python
data.isnull().sum()
```

```python
x_train,x_test,y_train,y_test = train_test_split(data.drop(columns=['has_covid']),data['has_covid'],test_size=0.2,random_state=43)
```

```python
x_train.head()
```

```python
colum_transformer = ColumnTransformer(transformers=[('tf1',SimpleImputer(),['fever']),
('tf2',OneHotEncoder(drop='first',sparse_output=False,dtype=int),['gender','city']),
    ('tf3',OrdinalEncoder(categories=[['Mild','Strong']]),['cough'])]
    ,remainder='passthrough')
```

```python
x_data=colum_transformer.fit_transform(x_train)
```

```python
x_test = colum_transformer.transform(x_test)
```
